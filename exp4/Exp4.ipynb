```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("diabetes.csv")
df.head()
```

```python
#Select appropriate variables
glucose = df["Glucose"]
bmi = df["BMI"]
outcome = df["Outcome"]
print("\n Selected variables are:")
df[['Glucose','BMI','Outcome']].head()
```

```python
# Divide patients into diabetic and non-diabetic groups

diabetic_group = df[df["Outcome"]==1]
non_diabetic_group = df[df["Outcome"]==0]
print("\n Number of diabetic patients:",len(diabetic_group))
print("Number of non-diabetic patients:",len(non_diabetic_group))
```

```python
# Display the groups
print("\nDiabetic Group:")
print(diabetic_group[["Glucose", "BMI", "Outcome"]].head())
print("\nNon-Diabetic Group:")
print(non_diabetic_group[["Glucose", "BMI", "Outcome"]].head())
```

```python
# Point Estimation
# Sample mean of Glucose
sample_mean_glucose = glucose.mean()

print("\n--- Point Estimation ---")
print("Sample Mean of Glucose:",
      sample_mean_glucose)
```

```python
# Sample Proportion of Diabetic Patients
diabetic_count = (df["Outcome"] == 1).sum()
total_patients = len(df)

sample_proportion = diabetic_count / total_patients

print("Number of diabetic patients:",
      diabetic_count)

print("Total number of patients:",
      total_patients)

print("Sample Proportion of Diabetic Patients:",
      sample_proportion)

print("Percentage of Diabetic Patients:",
      sample_proportion * 100)
```

```python
# Confidence interval for mean glucose
n = len(glucose)
mean_glucose = glucose.mean()
std_glucose = glucose.std()

confidence_level = 0.95
alpha = 1 - confidence_level

# Critical t-value
t_critical = stats.t.ppf(
    1 - alpha / 2,
    df=n - 1
)

# Standard error
standard_error = std_glucose / np.sqrt(n)

# Margin of error
margin_of_error = t_critical * standard_error

# Confidence interval
lower_limit = mean_glucose - margin_of_error
upper_limit = mean_glucose + margin_of_error

print("\n--- 95% Confidence Interval ---")
print("Sample Mean:", mean_glucose)
print("Standard Error:", standard_error)
print("Margin of Error:", margin_of_error)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

print(
    f"\n95% Confidence Interval for population mean glucose: "
    f"({lower_limit:.2f}, {upper_limit:.2f})"
)
```

```python
# One sample t-test
# Research Question:
# Is the mean glucose level significantly different from 120?

# H0: Population mean glucose = 120
# H1: Population mean glucose != 120

reference_value = 120

t_statistic, p_value = stats.ttest_1samp(
    glucose,
    reference_value
)

print("\n--- One-Sample t-Test ---")
print("Reference Mean:", reference_value)
print("t-statistic:", t_statistic)
print("p-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Decision: Reject H0")
    print("Conclusion: The mean glucose level is significantly different from 120.")
else:
    print("Decision: Fail to Reject H0")
    print("Conclusion: There is insufficient evidence that the mean glucose level differs from 120.")
```

```python
# Two sample t-test
diabetic_bmi = diabetic_group["BMI"]
non_diabetic_bmi = non_diabetic_group["BMI"]

# H0: Mean BMI of both groups is equal
# H1: Mean BMI of both groups is different

t_statistic, p_value = stats.ttest_ind(
    diabetic_bmi,
    non_diabetic_bmi,
    equal_var=False
)

print("\n--- Two-Sample t-Test ---")

print("Mean BMI of Diabetic Patients:",
      diabetic_bmi.mean())

print("Mean BMI of Non-Diabetic Patients:",
      non_diabetic_bmi.mean())

print("t-statistic:", t_statistic)
print("p-value:", p_value)

if p_value < alpha:
    print("Decision: Reject H0")
    print("Conclusion: There is a significant difference in mean BMI between diabetic and non-diabetic patients.")
else:
    print("Decision: Fail to Reject H0")
    print("Conclusion: There is insufficient evidence of a difference in mean BMI between the two groups.")
```

```python
#Compare
alpha = 0.05

if p_value < alpha:
    print("Reject H0")
else:
    print("Fail to Reject H0")
```

```python
#Create BMI Categories
df["BMI_Category"] = np.where(
    df["BMI"] < 25,
    "Normal",
    "High"
)

df[["BMI", "BMI_Category", "Outcome"]].head()
```

```python
#Create table
table = pd.crosstab(
    df["BMI_Category"],
    df["Outcome"]
)

print(table)
```

```python
#Chi-Square Test
from scipy.stats import chi2_contingency

chi2, p_value, dof, expected = chi2_contingency(table)

print("Chi-Square Test")
print("Chi-Square Statistic:", chi2)
print("P-value:", p_value)
print("Degrees of Freedom:", dof)
```

```python
#Compare
alpha = 0.05

if p_value < alpha:
    print("Reject H0")
else:
    print("Fail to Reject H0")
```
